### **string_functions**

```python
print("Hello " + "10")
print("Hello\n" * 5)
```

```python
print("*****\n" * 3)
```

```python
# isalpha
a = "helloworld"
print(a.isalpha())
print("mani".isalpha())
```

```python
# isdigit
a = "12345678"
x = a.isdigit()
print(x)
```

```python
# isalnum
a = "1234Helloworld"
x = a.isalnum()
print(x)
```

```python
# islower()
a = "Hello 1123world"
x = a.islower()
print(x)
```

```python
# isupper
a = "HELLO WORLD"
x = a.isupper()
print(x)
```

```python
# upper()

a = "123Hello WORld"
x = a.upper()
print(x)
```

```python
# Lower()

a = 'HELLO WORLD'
x = a.lower()
print(x)
```

```python
# # ends with()

mylist = ["new.txt","new.py","hello.py","my.txt"]

for each in mylist:
    if each.endswith(".py"):
        print(each)
```

```python
# title

a = 'HELLO WORLD Its a GreaT EaRTH'
x = a.title()
print(x)
```

```python
# Capitalize()

a = "hellO WorlD"
x =a.capitalize()
print(x)
```

```python
#spilt

a = "onomatopoeiao"
x = a.split("o")
print(x)
```

```python
#Join()

a = ["Hello","world","I\'am","mani"]
x = " ".join(a)
print(x)
```
