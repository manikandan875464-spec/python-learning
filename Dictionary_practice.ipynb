```python
mydict = {
"Name":"mani",
"Age" : 22,
"City": "Cennai",
500 : 50.55,
}
print(mydict,type(mydict))
print(mydict[500])
mydict["City"]='Mumbai'
mydict["Course"]="Python"
print(mydict)
```

```python
# Keys()

mydict = {
"Name":"manikandan",
"Age" : 22,
"City": "Cennai",
500 : 50.55,
}
print(mydict)
print()
print(list(mydict.keys()))
print()
ans = list(mydict.keys())
print(ans[1])
print()
for each in mydict.keys():
    
    print(each)
```

```python
# values()

mydict = {
"Name":"mani",
"Age" : 22,
"City": "theni",
500 : 50.55,
}
print(mydict)
print(list(mydict.values()))
print()
for each in mydict.values():
    print(each)
```

```python
# items()

mydict = {
"Name":"ram",
"Age" : 60,
"City": "madurai",
500 : 50.55,
}
print(mydict)
print()
print(list(mydict.items()))
print()
for each in mydict.items():
    print(f"{each[0]}\t{each[1]}")
```

```python
# pop(keys)

mydict = {
"Name":"Ajith",
"Age" : 29,
"City": "erode",
500 : 50.55,
}
print(mydict)
mydict.pop("Age")
print(mydict)
```

```python
# popitem()

mydict = {
"Name":"mani",
"Age" : 34,
"City": "Cennai",
500 : 50.55,
}
print(mydict)
mydict.popitem()
mydict.popitem()
mydict.popitem()
print(mydict)
```

```python
# update()

mydict = {
"Name":"ranjith",
"Age" : 26,
"City": "trichy",
500 : 70.77,
}
print(mydict)
newdit = {"Name":"Mani","Age":22,"City":"Kovai","course":"Python"}
mydict.update(newdit)
print(mydict)
```
