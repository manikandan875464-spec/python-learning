### **while_loop**

```python
a = 1
while a<10:
    print(a, end=" ")
    a+=1
print("loop over")
```

```python
a = 0
while a<=200:
    print(a,end=" ")
    a+=20
    
print("loop over")
```

### **break**

```python
a = 1      # infinity loop
while a>0:
    print(a,end=" ")
    a+= 1
    
```

```python
a = 1
while a>0:
    if a ==5:
        print(a)
        break
    print(a,end= " ")
    a+=1
    
    
    
    
```

```python
a = 1
while a<=20:
    if a % 3 ==0:
        print(a,end= " ")
        a += 1
        continue
    a += 1
```

```python
a = 1
while a<=5:
    b = 1
    while b<5:
        print(b,end = "")
        b +=1
    a+= 1
    print()
```

### **for_loop**

```python
mystr = "hello world"
for x in mystr[5:]:
    print(x, end ="")
```

```python
mylist= [10,20.55,"hello"]
for x in mylist:
    print(x)
```

```python
for x in range(0,10+1):
    print(x,end=" ")
```

```python
num = int(input("enter a number"))
for x in range (num+1):
    print(x,end = " ")
```

```python
for x in range(0,10,2):
    print(x,end = " ")
```

```python
for x in range(10,0,-1):
    print(x,end = " ")
```

```python
for x in "helloworld":
    if x == ' ':
        break
    print(x,end="")
else:
    print("\n the above loop completed")
```

```python
mylist = [2,6,9,30,60,11,16,97]

for each in mylist:
    if each %5 ==0:
        print(each,"divisible by 5")
        break
    else:
        print("number divisible by 5 not found")
```
