### **if elif and else**

```python
score = 100
secret_number = 50
guess = int(input("enter your guess:"))


if guess == secret_number:
    print("correct")
    score = score + 20
    print(score,"your score:")
    
elif guess > secret_number:
    print("too high")
    score = score - 10
    print(score,"your score:")
    
    
else:
    print("too low!")
    score = score - 10
    print(score,"your score:")
    
    
```
