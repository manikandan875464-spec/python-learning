### **escape sequence**

* \ - for newline in editor
* \\ - prints \
* \’ – prints ‘
* \” – prints “
* \n – prints a new line
* \t – prints a tab

```python
print("hello\\world\\")
print("10\\100")
print('I\'am python world')
print("Hello\"world\"")
print("hello\nWorld")
print("hello\tworld")
```
