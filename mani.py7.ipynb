### **function with parameter**

```python
def myfunc():
    x=100
    y=300
    z=x+y
    print("the sum of =",z)
    
    
myfunc()
```

```python
def myfunc(x,y):
    print("x=",x)
    print("y=",y)
    print("the sum of =",x+y)
    
myfunc(20,30)

print()
```

### **function with default parameter**

```python
def myfunc(x=0,y=0):
    print("x=",x)
    print("y=",y)
    print("the sum of =",x+y)
    
myfunc()
    
```

### **function with return keyword**

```python
def myfunc(x=0,y=0):
    print("x=",x)
    print("y=",y)
    print("the sum of =",x+y)
    return x+y
print()
ans = myfunc(100,200)
print(ans)
print(myfunc(400,50))
    
```
